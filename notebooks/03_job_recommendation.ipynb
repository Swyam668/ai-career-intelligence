```python
import pandas as pd

df = pd.read_csv("../data/raw/job_descriptions.csv", nrows=25000)
```

```python
features = [
    "Job Title",
    "Role",
    "Skills",
    "Experience",
    "Qualifications",
    "Location",
    "Work Type",
    "Job Description",
    "Responsibilities",
    "Salary Range"
]
```

```python
df["job_profile"] = (
    df["Job Title"].fillna("") + " " +
    df["Role"].fillna("") + " " +
    df["skills"].fillna("") + " " +
    df["Qualifications"].fillna("") + " " +
    df["Experience"].astype(str).fillna("") + " " +
    df["Job Description"].fillna("")
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=5000
)

job_vectors = vectorizer.fit_transform(df["job_profile"])
```

```python
#from resume parsing
user_profile = """
Python SQL Pandas
Machine Learning
Bachelor Degree
2 years experience
"""

user_profile1 = """
Javascript SQL C++
Web development
Bachelor Degree
2 years experience
"""
```

```python
user_vector = vectorizer.transform([user_profile])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

scores = cosine_similarity(
    user_vector,
    job_vectors
).flatten()

top_jobs = scores.argsort()[-10:][::-1]
```

```python
df["job_text"] = (
    df["skills"].fillna("") + " " +
    df["Job Description"].fillna("") + " " +
    df["Responsibilities"].fillna("")
).str.lower()
```

```python
import re

def extract_skills(text):
    text = text.lower()

    found = set()

    for skill in KNOWN_SKILLS:
        pattern = r"\b" + re.escape(skill.lower()) + r"\b"

        if re.search(pattern, text):
            found.add(skill.lower())

    return found
```

```python
df["extracted_skills"] = df["job_text"].apply(
    extract_skills
)
```

```python
def match_percentage(
    user_skills,
    job_skills
):

    if not job_skills:
        return 0

    return round(
        len(user_skills & job_skills)
        / len(job_skills)
        * 100,
        2
    )
```

```python

# the skill-gap analysis report
for idx in top_jobs[:5]:

    row = df.iloc[idx]

    job_skills = row["extracted_skills"]

    matched = user_skills & job_skills

    missing = job_skills - user_skills

    score = match_percentage(
        user_skills,
        job_skills
    )

    print("\n----------------")

    print("Job:", row["Job Title"])

    print("Match:", score, "%")

    print("Matched Skills:")
    print(sorted(matched))

    print("Missing Skills:")
    print(sorted(missing))
```

```python
results = []

for idx in top_jobs:

    row = df.iloc[idx]

    job_skills = row["extracted_skills"]

    matched = user_skills & job_skills

    missing = job_skills - user_skills

    score = match_percentage(
        user_skills,
        job_skills
    )

    results.append({
        "Job Title": row["Job Title"],
        "Similarity Score": scores[idx],
        "Match Percentage": score,
        "Matched Skills": ", ".join(sorted(matched)),
        "Missing Skills": ", ".join(sorted(missing))
    })
```
